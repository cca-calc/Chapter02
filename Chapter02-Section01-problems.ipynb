### Various expressions we will see for the slope of a function at a point.
- in this program, the last formula is used in the graphing.

<div style="padding-left: 5em; text-align: left; font-size: 1.5em;" >

$m = \frac{y_2 - y_1}{x_2 - x_1}$

$m = \frac{f(x_2) - f(x_1)}{x_2 - x_1}$

$m = \frac{f(x) - f(a)}{x - a}$

$m = \frac{f(b) - f(a)}{b - a}$

$m = \frac{f(x+h) - f(x)}{h}$

$m = \frac{f(x+h) - f(x-h)}{2h}$

```python


x = np.array([0, 5, 10, 15, 20, 25, 30])
y = np.array([1000, 694, 444, 250, 111, 28, 0])

plt.figure(figsize=(8, 5))
plt.plot(x, y, marker='o', linestyle='-', color='steelblue')

plt.title("Problem #1")
plt.xlabel("Time (min)")
plt.ylabel("Water (gallons)")
plt.grid(True)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Fixed point P
Px, Py = 15, 250

# Compute secant slopes for all Q != P
secants = []
for xi, yi in zip(x, y):
    if xi == Px:
        continue  # skip P itself
    slope = (yi - Py) / (xi - Px)
    secants.append((xi, yi, slope))

# Print table of slopes
print("Secant slopes using P(15,250):\n")
for xi, yi, slope in secants:
    print(f"Q = ({xi:.3f}, {yi:.3f})   slope m = {slope:.3f}")


```

```python
import numpy as np
import matplotlib.pyplot as plt

x2 = np.linspace(1.5, 2.5, 400)

def f(x):
    return 1/(1 - x)

plt.figure(figsize=(10,6))
plt.plot(x2, f(x2),color='green')


plt.title("f(x) = 1/(1 - x)")
plt.xlabel("x")
plt.ylabel("f(x)")
plt.grid(True)
plt.show()
```

```python
# Given x-values
x = np.array([1.5, 1.9, 1.99, 1.999, 2.5, 2.1, 2.01, 2.001])

# Function definition
def f(x):
    return 1/(1 - x)

# Compute y-values
y = f(x)

# Fixed point P
Px, Py = 2, -1

# Compute secant slopes for all Q
secants = []
for xi, yi in zip(x, y):
    slope = (yi - Py) / (xi - Px)
    secants.append((xi, yi, slope))

# Print table of slopes
print("Secant slopes using P(2, -1):\n")
for xi, yi, slope in secants:
    print(f"Q = ({xi:.3f}, {yi:.3f})   slope m = {slope:.6f}")

```

```python
t = np.array([0, 1, 2, 3, 4, 5, 6])
x = np.array([0, 4.9, 20.6, 46.5, 79.2, 124.8, 176.7])
plt.figure(figsize=(8, 5))
plt.plot(t, x, marker='o', linestyle='-', color='steelblue')

plt.title("Problem #1")
plt.xlabel("Time (sec)")
plt.ylabel("Position (meters?)")
plt.grid(True)
plt.show()
```

```python
# Compute average velocities for each interval
avg_vel = (x[1:] - x[:-1]) / (t[1:] - t[:-1])

# Print results
print("Average velocity on each interval:")
for i in range(len(avg_vel)):
    print(f"[{t[i]}, {t[i+1]}]  ->  {avg_vel[i]:.3f} m/s")

# Plot average velocity vs interval index
plt.figure(figsize=(10,5))
plt.plot(range(len(avg_vel)), avg_vel, marker='o')
plt.title("Average Velocity on Each Time Interval")
plt.xlabel("Interval index (i → interval [t_i, t_{i+1}])")
plt.ylabel("Average velocity (m/s)")
plt.grid(True)
plt.show()
```
