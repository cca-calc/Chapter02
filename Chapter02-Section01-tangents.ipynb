### Chapter 2, Section 1: Tangent and Velocity problems
- The first plot is of a parabola and it's tangent lines.
- The second plot is same using x**3.
- The third plot is same using x**4.
- The values of the slope of the tangents can be computed: If $f(x) = x^n$ then the slope of the tangent is $m=nx^{n-1}$.
- Problems 1, 3, 7

```python
import numpy as np
import matplotlib.pyplot as plt
```

### Various expressions we will see for the slope of a function at a point.
- in this program, the last formula is used in the graphing.

<div style="padding-left: 5em; text-align: left; font-size: 1.5em;" >

$m = \frac{y_2 - y_1}{x_2 - x_1}$

$m = \frac{f(x_2) - f(x_1)}{x_2 - x_1}$

$m = \frac{f(x) - f(a)}{x - a}$

$m = \frac{f(b) - f(a)}{b - a}$

$m = \frac{f(x+h) - f(x)}{h}$

$m = \frac{f(x+h) - f(x-h)}{2h}$

</div>

### Terms to know:


```python

# Define x values
a = 5
x = np.linspace(-a, a, 400)
y = x**2

# Points where the tangent lines are computed.
ntan = 5
x_points = np.linspace(-a, a, ntan)
y_points = x_points**2

# Using the slope formula rise/run or (change in y)/(change in x)
delta = 1.0
slopes = ((x_points + delta)**2 - (x_points - delta)**2)/(2*delta)

# Set up subplots
fig, axs = plt.subplots(3, 1, figsize=(10, 10))

# Plot x*x and tangent lines
axs[0].plot(x, y, label='x*x', color='black')
for x0, y0, m in zip(x_points, y_points, slopes):
    x_tangent = np.linspace(x0 - delta, x0 + delta, 10)
    y_tangent = m * (x_tangent - x0) + y0
    axs[0].plot(x_tangent, y_tangent, color='red', linewidth=1)
axs[0].set_title('x*x and Tangent Lines')
axs[0].set_xlabel('x')
axs[0].set_ylabel('y')
axs[0].grid(True)
axs[0].legend()

for x0, y0, m in zip(x_points, y_points, slopes):
    x_tangent = np.linspace(x0 - delta, x0 + delta, 10)
    y_tangent = m * (x_tangent - x0) + y0
    axs[1].plot(x_tangent, y_tangent, color='red',  linewidth=1)
axs[1].set_title('Tangent Lines')
axs[1].set_xlabel('x')
axs[1].set_ylabel('y')
axs[1].grid(True)

# Plot slope values
axs[2].plot(x_points, slopes, 'go-', label='Slope = 2x', markersize=6)
axs[2].set_title('Slope of Tangents to x*x')
axs[2].set_xlabel('x')
axs[2].set_ylabel('Slope')
axs[2].grid(True)
axs[2].legend()

plt.tight_layout()
plt.show()
```
