```python
import numpy as np
import matplotlib.pyplot as plt
```

```python

# Define x values, compute f(x)
a = 5
x_right = np.linspace(-0.99, a, 400)
x_left = np.linspace(-a, -1.01, 400)
y_right = ((x_right-1)/(x_right*x_right - 1))
y_left = ((x_left-1)/(x_left*x_left - 1))

# Plot f(x)
plt.plot(x_right, y_right, label=r'$\frac{x-1}{x^2 -1}$', color='red')
plt.plot(x_left, y_left, color='green')
plt.title('Function')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
# Define x values, compute f(x)
x = np.linspace(-1.05, 1.05, 100)
y = ((x*x+9)**0.5-3)/(x*x)

# Plot f(x)
plt.plot(x, y, label=r'$\frac{\sqrt{t^2 + 9} - 3}{t^2}$', color='black')
plt.title('Function')
plt.xlabel('x')
plt.ylabel('f(x)')
plt.grid(True)
plt.legend()
plt.show()
```
